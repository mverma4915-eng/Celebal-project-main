## Import Libraries

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
from transformers import pipeline

print("All imports successful!")
```

```python
import os

print(os.listdir())
```

```python
loader = PyPDFLoader("Week7_Project.pdf")

documents = loader.load()

print(documents[0].page_content)
```

## Split the Text into Chunks

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(
    documents
)

print(
    "Number of Chunks:",
    len(chunks)
)
```

## Convert Chunks into Text

```python
texts = [
    chunk.page_content
    for chunk in chunks
]

print(texts[0])
```

## Create Embeddings

```python
model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

embeddings = model.encode(
    texts
)

print(
    embeddings.shape
)
```

## Create FAISS Vector Database

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(
    dimension
)

index.add(
    np.array(embeddings)
)

print(
    "Embeddings Stored Successfully!"
)
```

## Ask a Question

```python
query = """
What are the skills
mentioned in the document?
"""
```

## Convert Question to Embedding

```python
query_embedding = model.encode(
    [query]
)
```

## Retrieve Top 3 Chunks

```python
k = 3

distances, indices = index.search(
    np.array(query_embedding),
    k
)

print(indices)
```

## Display Retrieved Chunks

```python
retrieved_chunks = []

for idx in indices[0]:
    retrieved_chunks.append(
        texts[idx]
    )

for chunk in retrieved_chunks:
    print(chunk)
    print("-"*50)
```

Load the Language Model

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="google/flan-t5-base"
)
```

## Prompt

```python
context = "\n".join(
    retrieved_chunks
)

prompt = f"""
Context:
{context}

Question:
{query}

Answer:
"""
```

## Final Answer

```python
answer = generator(
    prompt,
    max_length=100
)

print(
    answer[0]["generated_text"]
)
```

## ## Conclusion

This project successfully developed and implemented a Retrieval-Augmented Generation (RAG) system that combines document retrieval and natural language generation to provide accurate and context-aware answers from custom PDF documents.
