# 🧠 # Smart Query Routing Agent System using Python




##  Project Objective

The objective of this project is to develop a simple Single Agent System that can intelligently route user queries to the appropriate tool based on the query type. The agent performs mathematical calculations, extracts keywords from text, and handles general queries while providing structured JSON outputs.

## Problem Statement

In modern AI systems, agents are required to understand user intent and select the appropriate tool to solve a problem. This project demonstrates how an agent can use conditional routing to perform different tasks efficiently while maintaining a simple and scalable architecture.

```python
# Import Libraries

import json
```

```python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
```

```python
# TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
```

## Agent Routing Logic

The agent follows the below logic:

1. If the query contains "calculate", the Calculator Tool is used.
2. If the query contains "keywords", the Keyword Extraction Tool is used.
3. Otherwise, the agent returns a general response.
4. The output is returned in JSON format.

```python
# YOUR CODE HERE

def agent(query: str):

    query_lower = query.lower()

    try:

        # Calculator Tool
        if "calculate" in query_lower:

            expression = query_lower.replace(
                "calculate", ""
            ).strip()

            result = calculator(expression)

            return {

                "type": "calculation",
                "result": result

            }

        # Keyword Extraction Tool
        elif "keywords" in query_lower:

            text = query.replace(
                "Extract keywords from", ""
            ).strip()

            result = extract_keywords(text)

            return {

                "type": "keywords",
                "result": result

            }

        # General Response
        else:

            return {

                "type": "general",
                "result": f"You asked: {query}"

            }

    # Error Handling
    except Exception as e:

        return {

            "type": "error",
            "result": str(e)

        }
```

## Error Handling

The project uses try-except blocks to handle unexpected inputs and calculation errors. If an exception occurs, the agent returns a structured JSON error response instead of crashing.

```python
# TEST CASES

queries = [

    "Calculate 20 + 5",

    "Extract keywords from Artificial Intelligence is transforming industries",

    "What is machine learning?"

]
```

```python
for q in queries:

    print("Query :", q)
    print("Response :", agent(q))
    print("-" * 50)
```
